```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
# True answer is w=3, b=4 
# Visualize the data
plt.scatter(X, y, alpha=0.6)
plt.title('Synthetic data true line is y=4+3x') # Add title to the plot
plt.xlabel('X');
plt.ylabel('y');
plt.show()
```

```python
# y_pred h(theta) = w*X + b will be the slope and b will be the intercept
# Initialize parameters
import numpy as np;
import matplotlib.pyplot as plt;
w = np.random.randn(1)  # weight (slope)
b = np.random.randn(1)  # bias (intercept)
learning_rate = 0.01
n_iterations = 1000
# error
#h(theta)-y(the true value)
for i in range(n_iterations):
    y_pred = w * X + b  # predicted values
    error = y_pred - y  # error
    loss = (error ** 2).mean()  # mean squared error
    dw = (2 / len(X)) * np.dot(X.T, error)  # gradient with respect to w
    db = (2 / len(X)) * np.sum(error)  # gradient with respect to b
    w -= learning_rate * dw  # update w
    b -= learning_rate * db  # update b
# loss
print(f"Final parameters: w={w[0][0]:.2f}, b={b[0][0]:.2f}")
print(f"Final loss: {loss:.2f}")
# dw
#dw is the gradient of the loss function with respect to the weight w.
#It indicates how much the loss would change if we were to change w by a small amount. 
#A positive dw means that increasing w would increase the loss, while a negative dw means that increasing w would decrease the loss. 
#The magnitude of dw indicates how steep the slope of the loss function is at the current value of w.
# db is the gradient of the loss function with respect to the bias b.
# update the parameters w and b by moving in the direction opposite to the gradient (since we want to minimize the loss).

# Visualize the fitted line
plt.scatter(X, y, alpha=0.6, label='Data')
x_line = np.linspace(0, 2, 100).reshape(-1, 1)
y_line = w * x_line + b
plt.plot(x_line, y_line, color='red', label='Fitted line')
plt.title('Linear Regression Fit') # Add title to the plot
plt.xlabel('X');
plt.ylabel('y');
plt.legend()
plt.show()
#practice only
```
